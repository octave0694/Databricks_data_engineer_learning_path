### Create Streaming Tables for incremental Processing

```sql
SELECT *
FROM read_files(
  '/Volumes/dbacademy/csv_files_autoloader_source',
  format => 'CSV',
  sep => '|',
  header => 'true'
);
```

```sql
CREATE OR REFRESH STREAMING TABLE sql_csv_autoloader
SCHEDULE EVERY 1 WEEK
AS
SELECT *
FROM STREAM read_files(
  '/Volumes/dbacademy/csv_files_autoloader_source',
  format => 'CSV',
  sep => '|',
  header => 'true'
);
```

```sql
SELECT *
FROM sql_csv_autoloader;
```

```sql
DESCRIBE TABLE EXTENDED sql_csv_autoloader;
```

```sql
DESCRIBE HISTORY sql_csv_autoloader;
```

After loading new file we can refresh the streaming table manually

```sql
REFRESH STREAMING TABLE sql_csv_autoloader;
```
